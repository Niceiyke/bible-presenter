# Bible Presenter — Asset Aggregator

This notebook downloads all required Bible database and AI model files and saves them to your Google Drive. 

**Steps:**
1. Mount your Google Drive.
2. Run all cells.
3. Go to the `BiblePresenterAssets` folder in your Google Drive.
4. Right-click each file → **Get link** → Set to **Anyone with the link**.
5. Copy the File IDs and update the Rust `model_manager.rs` constants.

```python
from google.colab import drive
import os
import requests

drive.mount('/content/drive')

ASSET_DIR = '/content/drive/MyDrive/BiblePresenterAssets'
os.makedirs(ASSET_DIR, exist_ok=True)
os.makedirs(os.path.join(ASSET_DIR, 'reranker'), exist_ok=True)

files_to_download = [
    # Bible DB
    ("https://raw.githubusercontent.com/alshival/wordlyte_bible/main/SUPER_BIBLE/wordlyte_bible.db", "wordlyte_bible.db"),
    
    # BGE Embedding Model
    ("https://huggingface.co/Xenova/bge-small-en-v1.5/resolve/main/onnx/model.onnx", "bge-small-en-v1.5.onnx"),
    ("https://huggingface.co/Xenova/bge-small-en-v1.5/resolve/main/tokenizer.json", "tokenizer.json"),
    
    # Reranker Model
    ("https://huggingface.co/Xenova/ms-marco-MiniLM-L-6-v2/resolve/main/onnx/model.onnx", "reranker/model.onnx"),
    ("https://huggingface.co/Xenova/ms-marco-MiniLM-L-6-v2/resolve/main/tokenizer.json", "reranker/tokenizer.json"),
]

def download_file(url, local_filename):
    path = os.path.join(ASSET_DIR, local_filename)
    if os.path.exists(path):
        print(f"Skipping {local_filename}, already exists.")
        return
    
    print(f"Downloading {local_filename}...")
    with requests.get(url, stream=True) as r:
        r.raise_for_status()
        with open(path, 'wb') as f:
            for chunk in r.iter_content(chunk_size=8192):
                f.write(chunk)
    print(f"Done: {local_filename}")

for url, name in files_to_download:
    try:
        download_file(url, name)
    except Exception as e:
        print(f"Failed to download {name}: {e}")

print("\nAll downloads completed. Files are in your Google Drive: /MyDrive/BiblePresenterAssets")
```
